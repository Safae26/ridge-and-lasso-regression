**Objective**: Build a predictive model capable of estimating a baseball player's salary based on their performance features and career statistics.

## Relevant Dependencies

```python
import pandas as pd # For data analysis
import numpy as np # Arrays manipulation
import math
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()

import warnings
warnings.filterwarnings("ignore")
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, RidgeCV, LassoCV
from sklearn.model_selection import RepeatedKFold, train_test_split
from sklearn.metrics import mean_squared_error
```

## Loading Data

```python
df = pd.read_csv("Hitters.csv")
df_hitters = df.copy() # to safe keep a version
df_hitters
```

## Dataset Description

- **322 players** with **20 features** (baseball terms, qualities and characteristics)
- **HmRuns**: Home runs made by a particular player in 1986

**Features starting with 'C'** correspond to an individual player's achievements throughout their entire baseball career:
- **CHits**: total number of hits
- **CRuns**: total number of runs
- etc...

**Target variable**: Salary (earnings of a player)

## Preprocessing

```python
# Categorical variables
print("The league types are : ", df_hitters["League"].unique())
print("The division types are : ", df_hitters["Division"].unique())
print("The new league types are : ", df_hitters["NewLeague"].unique())
```

```python
# Converting categorical variables into 0s and 1s
df_hitters_num = pd.get_dummies(df_hitters, columns = ['League', 'Division', 'NewLeague'], drop_first=True)
df_hitters_num
```

```python
# Checking null values
df_hitters_num.isnull().sum()
```

**Missing Values Strategy**: Drop samples with null values initially, then use predicted values from the trained model to fill them later.

```python
df_hitters_num_nonull = df_hitters_num.dropna()
```

```python
df_hitters_num_nonull.isnull().sum()
```

## Exploratory Data Analysis

```python
sns.displot(df_hitters_num_nonull['Salary']);
```

```python
correlation = df_hitters_num_nonull.corr()
correlation["Salary"].sort_values()
```

**Pearson's Correlation**: Coefficient values range from -1 to 1:
- **1**: Strong positive relationship
- **-1**: Strong negative relationship  
- **0**: No relationship

```python
# Checking for multicolinearity (correlation between independent variables)
plt.figure(figsize=(13, 6))
sns.heatmap(df_hitters_num_nonull.corr(),
             vmin=-1,
             vmax=1,
            
             cmap="GnBu",
             annot=True
            )
plt.show()
```

## Declaring Dependent and Independent Variables

```python
X = df_hitters_num_nonull.drop("Salary", axis=1)
y = df_hitters_num_nonull["Salary"]
```

## Splittinig the Data into train and test sets

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.35, random_state=365)
```

**Feature Scaling**: Standardize training and testing data to ensure all variables contribute equally to the algorithm.

```python
scaler = StandardScaler() 
```

```python
# For train data
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test) 
```

**StandardScaler**: Transforms data distribution to mean=0, std=1 for train set, then applies the same scaling metrics to the test set.

## Performing Linear Regression

```python
lin_reg = LinearRegression()
lin_reg.fit(X_train, y_train)
```

```python
print("Linear Regression coefficients : ", lin_reg.coef_)
print("Linear Regression y-intercept : ", lin_reg.intercept_)
```

**High coefficient : more emphasis on a particular feature (independant variable)**

```python
# Predictions
lin_reg_y_pred = lin_reg.predict(X_test)
lin_reg_y_pred
```

```python
# Comparison
lin_comp = pd.DataFrame({'Predicted':lin_reg_y_pred, 'Actual': y_test})
lin_comp
```

**Evaluation Metrics**:
- **MSE**: Mean Squared Error
- **R²**: Coefficient of determination (0-1) measuring relationship strength between dependent and independent variables

```python
print("Linear Regression Model RMSE : ", math.sqrt(mean_squared_error(y_test, lin_reg_y_pred))) # The lesser the better
print("Linear Regression Model Training Score : ", lin_reg.score(X_train, y_train))
print("Linear Regression Model Testing Score : ", lin_reg.score(X_test, y_test))
```

## Performing Ridge Regression

**Hyperparameter Tuning**: Use cross-validation to fine-tune lambda (regularization parameter)

```python
# A well chosen lambda to control the penalty term
cv = RepeatedKFold(n_splits=5, n_repeats=3, random_state=1)

ridge = RidgeCV(alphas=np.arange(0.1, 10, 0.1), cv=cv, scoring="neg_mean_squared_error") # lambda is alpha in sklearn
```

**RidgeCV Scoring**: The scoring parameter evaluates tuning parameter effectiveness for optimal performance

```python
ridge.fit(X_train, y_train)

ridge_reg_y_pred = ridge.predict(X_test)
ridge_reg_y_pred
```

```python
print("Ridge tuning  parameter : ", ridge.alpha_)
print("Ridge model coefficients : ", ridge.coef_)
print("Ridge model y-intercept : ", ridge.intercept_)
```

```python
print("Ridge Regression Model RMSE : ", math.sqrt(mean_squared_error(y_test, ridge_reg_y_pred))) 
print("Ridge Regression Model Training Score : ", ridge.score(X_train, y_train))
print("Ridge Regression Model Testing Score : ", ridge.score(X_test, y_test))
```

## Performing Lasso Regression

```python
lasso = LassoCV(alphas=np.arange(0.1, 10, 0.1), cv=cv, tol=1) 
```

**LassoCV Tolerance**: 
- `tol`: Optimization tolerance - stops search when tolerance between min and max is achieved
- `tol=1`: Prevents duality gap issues

```python
lasso.fit(X_train, y_train)

lasso_reg_y_pred = lasso.predict(X_test)
lasso_reg_y_pred
```

```python
print("Lasso tuning  parameter : ", lasso.alpha_)
print("Lasso model coefficients : ", lasso.coef_)
print("Lasso model y-intercept : ", lasso.intercept_)
```

```python
print("Lasso Regression Model RMSE : ", math.sqrt(mean_squared_error(y_test, lasso_reg_y_pred))) 
print("Lasso Regression Model Training Score : ", lasso.score(X_train, y_train))
print("Lasso Regression Model Testing Score : ", lasso.score(X_test, y_test))
```

**Model Comparison**: After evaluating all three models, Ridge and Lasso regressions demonstrate better performance than Linear Regression 🏆
